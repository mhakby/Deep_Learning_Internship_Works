```python
# Jupyter Notebook - Pre-Feature Engineering
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, LabelEncoder
```

* With pandas we will import and manipulate data.
* With OneHotEncoder and LabelEncoder we will convert non-numeric values to numeric.

```python
# Load the dataset
data = pd.read_csv('../data/data.csv')
```

```python
# Initialize LabelEncoder for ordinal encoding
label_encoder = LabelEncoder()
data['Category'] = label_encoder.fit_transform(data['Category'])
```

```python
# Initialize OneHotEncoder for nominal encoding
one_hot_encoder = OneHotEncoder(sparse_output=False)
```

```python
# Applying One-Hot Encoding to the 'Color' feature
encoded_features = one_hot_encoder.fit_transform(data[['Color']])
```

```python
# Convert the encoded features to a DataFrame for better visualization
encoded_df = pd.DataFrame(encoded_features, columns=one_hot_encoder.get_feature_names_out(['Color']))
```

```python
# Concatenate the original dataset with the encoded features
data = pd.concat([data, encoded_df], axis=1).drop(['Color'], axis=1)
```

```python
# Display the first few rows of the transformed dataset
print(data.head())
```

I used LabelEncoder for the original categorical feature:
* 0 for Clothing
* 1 for Electronics
* 2 for Furniture

Also I used OneHotEncoder for the 'Color' feature:
* 1.0 indicates the presence of that color, and 0.0 indicates its absence.
